```python
import boto3
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Fetch credentials from environment variables
aws_access_key = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_key = os.getenv("AWS_SECRET_ACCESS_KEY")

# Define S3 bucket and prefix
BUCKET_NAME = "troyschwab-resources"
PREFIX = "photography/midres/"

# Set up the S3 client with credentials
s3 = boto3.client(
    "s3",
    aws_access_key_id=aws_access_key,
    aws_secret_access_key=aws_secret_key
)

def download_images():
    # List all objects in the specified path
    response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix=PREFIX)

    if "Contents" not in response:
        print("No images found.")
        return

    # Create a local directory to store images
    os.makedirs("downloaded_images", exist_ok=True)

    # Download each image
    for obj in response["Contents"]:
        key = obj["Key"]
        filename = key.split("/")[-1]  # Extract the filename from the path
        local_path = os.path.join("downloaded_images", filename)

        print(f"Downloading {filename}...")
        s3.download_file(BUCKET_NAME, key, local_path)

    print("Download complete.")

if __name__ == "__main__":
    download_images()
```
